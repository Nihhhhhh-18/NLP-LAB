**TASK 1**

```python
import nltk
from nltk.corpus import brown

nltk.download('brown')

# Load Brown Corpus
sentences = brown.sents()

# Select first 20 sentences
selected_sentences = sentences[:20]

print("Number of sentences:", len(selected_sentences))

for i, sentence in enumerate(selected_sentences, 1):
    print(f"Sentence {i}: {' '.join(sentence)}")
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

# Convert selected sentences into one text
text = " ".join([" ".join(sentence) for sentence in selected_sentences])

# Sentence tokenization
sentence_tokens = sent_tokenize(text)

print("Sentence Tokens:")
for i, sentence in enumerate(sentence_tokens, 1):
    print(f"{i}: {sentence}")

# Word tokenization
print("\nWord Tokens:")
for i, sentence in enumerate(sentence_tokens[:5], 1):
    print(f"\nSentence {i}:")
    print(word_tokenize(sentence))
```

```python
# Convert all word tokens to lowercase

lowercase_tokens = []

for sentence in sentence_tokens:
    words = word_tokenize(sentence)
    lower_words = [word.lower() for word in words]
    lowercase_tokens.append(lower_words)

print("Lowercase Tokens:")

for i, tokens in enumerate(lowercase_tokens[:5], 1):
    print(f"Sentence {i}:")
    print(tokens)
```

```python
import string
from nltk.corpus import stopwords

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

cleaned_tokens = []

for tokens in lowercase_tokens:
    cleaned = [
        word for word in tokens
        if word not in string.punctuation and word not in stop_words
    ]
    cleaned_tokens.append(cleaned)

print("Cleaned Tokens:")

for i, tokens in enumerate(cleaned_tokens[:5], 1):
    print(f"Sentence {i}:")
    print(tokens)
```

```python
print("Original and Cleaned Tokens")
print("=" * 70)

for i in range(5):

    original = word_tokenize(sentence_tokens[i])

    print(f"\nSentence {i+1}:")
    print("Original Tokens :", original)
    print("Cleaned Tokens  :", cleaned_tokens[i])
```

**TASK 2**

```python
from nltk.corpus import brown

# Extract alphabetic words from Brown Corpus
alphabetic_words = [
    word.lower()
    for word in brown.words()
    if word.isalpha()
]

# Select 30 words
selected_words = alphabetic_words[:30]

# Add required words from the question
selected_words.extend([
    "playing",
    "studies",
    "running",
    "better",
    "cars"
])

print("Words:")
print(selected_words)
print("\nTotal words:", len(selected_words))
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in selected_words
]

print("Original -> Stemmed")
print("=" * 40)

for original, stemmed in zip(selected_words, stemmed_words):
    print(f"{original:15} -> {stemmed}")
```

```python
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in selected_words
]

print("Original -> Lemmatized")
print("=" * 40)

for original, lemma in zip(selected_words, lemmatized_words):
    print(f"{original:15} -> {lemma}")
```

```python
required_words = [
    "playing",
    "studies",
    "running",
    "better",
    "cars"
]

print("Required Words Comparison")
print("=" * 50)

for word in required_words:
    stem = stemmer.stem(word)
    lemma = lemmatizer.lemmatize(word)

    print(f"{word:12} -> Stem: {stem:10} -> Lemma: {lemma}")
```

```python
import pandas as pd

df = pd.DataFrame({
    "Original": selected_words,
    "Stemmed": stemmed_words,
    "Lemmatized": lemmatized_words
})

display(df)
```

**TASK 3**

```python
from nltk.corpus import words

nltk.download('words')

# Load English vocabulary
word_list = words.words()

# Convert to lowercase
vocabulary = set(
    word.lower()
    for word in word_list
    if word.isalpha()
)

print("Total vocabulary words:", len(vocabulary))
```

```python
misspelled_words = [
    "recieve",
    "becuase",
    "definately",
    "seperate",
    "occured",
    "teh",
    "adress",
    "enviroment",
    "langauge",
    "studing"
]

print("Misspelled Words:")
for word in misspelled_words:
    print(word)
```

```python
import difflib

def correct_word(word):

    suggestions = difflib.get_close_matches(
        word.lower(),
        vocabulary,
        n=1,
        cutoff=0.6
    )

    if suggestions:
        return suggestions[0]

    return word


print("Spelling Corrections")
print("=" * 50)

for word in misspelled_words:
    correction = correct_word(word)
    print(f"{word:15} -> {correction}")
```

```python
print("Misspelled Word | Suggested Correction")
print("=" * 50)

for word in misspelled_words:
    suggestion = correct_word(word)
    print(f"{word:15} | {suggestion}")
```

```python
import re

sentences = [
    "I definately recieved the adress.",
    "The enviroment is very beutiful.",
    "I am studing langauge processing.",
    "This occured becuase of a mistake.",
    "Please seperate the files."
]

def correct_sentence(sentence):

    words_in_sentence = re.findall(r'\b[a-zA-Z]+\b', sentence)

    for word in words_in_sentence:
        correction = correct_word(word)

        sentence = re.sub(
            r'\b' + re.escape(word) + r'\b',
            correction,
            sentence,
            count=1
        )

    return sentence


for sentence in sentences:

    corrected = correct_sentence(sentence)

    print("\nOriginal:")
    print(sentence)

    print("Corrected:")
    print(corrected)
```

```python
from nltk.corpus import brown

nltk.download('brown')

# Load Brown Corpus
brown_words = brown.words()

# Select a suitable subset
subset = brown_words[:10000]

print("Number of words:", len(subset))
print(subset[:50])
```

```python
# Convert to lowercase and keep only alphabetic words

cleaned_words = [
    word.lower()
    for word in subset
    if word.isalpha()
]

print("Number of cleaned words:", len(cleaned_words))
print("\nFirst 50 cleaned words:")
print(cleaned_words[:50])
```

```python
from nltk.util import ngrams

# Generate N-grams

unigrams = list(ngrams(cleaned_words, 1))
bigrams = list(ngrams(cleaned_words, 2))
trigrams = list(ngrams(cleaned_words, 3))

print("First 10 Unigrams:")
print(unigrams[:10])

print("\nFirst 10 Bigrams:")
print(bigrams[:10])

print("\nFirst 10 Trigrams:")
print(trigrams[:10])
```

```python
from collections import Counter

# Calculate frequencies

unigram_frequency = Counter(unigrams)
bigram_frequency = Counter(bigrams)
trigram_frequency = Counter(trigrams)

print("Number of unique unigrams:", len(unigram_frequency))
print("Number of unique bigrams:", len(bigram_frequency))
print("Number of unique trigrams:", len(trigram_frequency))
```

```python
print("TOP 10 UNIGRAMS")
print("=" * 40)

for gram, frequency in unigram_frequency.most_common(10):
    print(gram, ":", frequency)


print("\nTOP 10 BIGRAMS")
print("=" * 40)

for gram, frequency in bigram_frequency.most_common(10):
    print(gram, ":", frequency)


print("\nTOP 10 TRIGRAMS")
print("=" * 40)

for gram, frequency in trigram_frequency.most_common(10):
    print(gram, ":", frequency)
```

```python
print("""
Unigram:
A unigram contains one word and provides very little context.

Bigram:
A bigram contains two consecutive words and captures the relationship
between neighboring words.

Trigram:
A trigram contains three consecutive words and provides more contextual
information than a unigram or bigram.

Therefore, increasing N generally captures more context, but larger
N-grams occur less frequently because exact word sequences become more
specific.
""")
```
