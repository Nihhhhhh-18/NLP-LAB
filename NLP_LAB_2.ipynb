```python

import nltk
import pandas as pd

from nltk.corpus import wordnet as wn
from nltk.corpus import brown


nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('brown')

```

```python

words = [
    "car",
    "bank",
    "bat",
    "dog",
    "light",
    "plant",
    "book",
    "teacher",
    "student",
    "computer"
]


for word in words:
    print("\n")
    print("WORD:", word)

    synsets = wn.synsets(word)

    if not synsets:
        print("No synsets found.")
        continue

    for synset in synsets:
        print("\nSynset:", synset.name())
        print("Definition:", synset.definition())

        examples = synset.examples()

        if examples:
            print("Example:", examples[0])
        else:
            print("Example: No example available")
```

```python
multiple_meaning_words = ["bank", "bat", "light"]

for word in multiple_meaning_words:
    print("\n" )
    print("MULTIPLE MEANINGS OF:", word)

    synsets = wn.synsets(word)

    for synset in synsets[:5]:
        print("\nSynset:", synset.name())
        print("Definition:", synset.definition())

        if synset.examples():
            print("Example:", synset.examples()[0])
```

```python
nouns = [
    "dog",
    "cat",
    "car",
    "apple",
    "teacher",
    "student",
    "computer",
    "book",
    "bird",
    "tree"
]


hypernym_data = []

for word in nouns:
    synsets = wn.synsets(word, pos=wn.NOUN)

    if not synsets:
        continue

    synset = synsets[0]


    direct_hypernyms = synset.hypernyms()

    hypernym_names = [h.name() for h in direct_hypernyms]

    hypernym_data.append({
        "Word": word,
        "Synset": synset.name(),
        "Definition": synset.definition(),
        "Direct Hypernyms": ", ".join(hypernym_names)
    })


hypernym_df = pd.DataFrame(hypernym_data)

print(hypernym_df.to_string(index=False))
```

```python
def get_hypernym_chain(synset, levels=3):
    chain = []
    current = synset

    for i in range(levels):
        hypernyms = current.hypernyms()

        if not hypernyms:
            break

        current = hypernyms[0]
        chain.append(current.name())

    return chain
```

```python
hierarchy_data = []

for word in nouns:
    synsets = wn.synsets(word, pos=wn.NOUN)

    if not synsets:
        continue

    synset = synsets[0]

    chain = get_hypernym_chain(synset, levels=3)

    hierarchy_data.append({
        "Word": word,
        "Synset": synset.name(),
        "Hypernym Level 1": chain[0] if len(chain) > 0 else "-",
        "Hypernym Level 2": chain[1] if len(chain) > 1 else "-",
        "Hypernym Level 3": chain[2] if len(chain) > 2 else "-"
    })

hierarchy_df = pd.DataFrame(hierarchy_data)

print(hierarchy_df.to_string(index=False))
```

```python
print("\nHYPERNYM HIERARCHY")

for word in nouns[:5]:
    synsets = wn.synsets(word, pos=wn.NOUN)

    if not synsets:
        continue

    synset = synsets[0]
    chain = get_hypernym_chain(synset, levels=3)

    print("\nWord:", word)
    print("Starting Synset:", synset.name())

    for i, hypernym in enumerate(chain, start=1):
        print("Level", i, ":", hypernym)
```

```python

word_pairs = [
    ("car", "automobile"),
    ("dog", "cat"),
    ("dog", "animal"),
    ("car", "vehicle"),
    ("apple", "fruit"),
    ("teacher", "student"),
    ("dog", "computer"),
    ("book", "tree")
]


selected_synsets = {
    "car": wn.synset("car.n.01"),
    "automobile": wn.synset("automobile.n.01"),
    "dog": wn.synset("dog.n.01"),
    "cat": wn.synset("cat.n.01"),
    "animal": wn.synset("animal.n.01"),
    "vehicle": wn.synset("vehicle.n.01"),
    "apple": wn.synset("apple.n.01"),
    "fruit": wn.synset("fruit.n.01"),
    "teacher": wn.synset("teacher.n.01"),
    "student": wn.synset("student.n.01"),
    "computer": wn.synset("computer.n.01"),
    "book": wn.synset("book.n.01"),
    "tree": wn.synset("tree.n.01")
}
```

```python

similarity_results = []

for word1, word2 in word_pairs:

    synset1 = selected_synsets[word1]
    synset2 = selected_synsets[word2]

    similarity = synset1.path_similarity(synset2)

    similarity_results.append({
        "Word 1": word1,
        "Word 2": word2,
        "Synset 1": synset1.name(),
        "Synset 2": synset2.name(),
        "Path Similarity": similarity
    })


similarity_df = pd.DataFrame(similarity_results)

print("SEMANTIC SIMILARITY RESULTS")


print(similarity_df.to_string(index=False))
```

```python

ranked_similarity = similarity_df.sort_values(
    by="Path Similarity",
    ascending=False
).reset_index(drop=True)


ranked_similarity.insert(
    0,
    "Rank",
    range(1, len(ranked_similarity) + 1)
)

print("\nRANKED SEMANTIC SIMILARITY")


print(ranked_similarity.to_string(index=False))
```
