```python

sentences = hindi_data[:500]

print("Number of sentences selected:", len(sentences))
```

```python
def preprocess_token(token):
    """
    Basic preprocessing:
    - Unicode normalization
    - Remove unwanted punctuation
    - Remove extra spaces
    """

    token = unicodedata.normalize("NFC", token)
    token = token.strip()


    if re.fullmatch(r'[^\w\u0900-\u097F]+', token):
        return ""

    return token


processed_tokens = []

for sentence in sentences:
    for token in sentence:
        token = preprocess_token(token)

        if token:
            processed_tokens.append(token)

print("Total processed tokens:", len(processed_tokens))

print("\nFirst 50 processed tokens:")
print(processed_tokens[:50])
```

```python

total_tokens = len(processed_tokens)


vocabulary = set(processed_tokens)
vocabulary_size = len(vocabulary)

print("Total Tokens:", total_tokens)
print("Vocabulary Size:", vocabulary_size)
```

```python

token_frequency = Counter(processed_tokens)


most_common_tokens = token_frequency.most_common(10)

print("10 Most Frequent Tokens:")
print()

for token, frequency in most_common_tokens:
    print(f"{token} -> {frequency}")
```

```python


indic_digit_map = str.maketrans(
    "०१२३४५६७८९",
    "0123456789"
)


def normalize_indic_text(text):
    """
    Normalizes Indic and mixed-language text.
    """


    text = unicodedata.normalize("NFC", text)


    text = text.translate(indic_digit_map)


    text = re.sub(r'\s+', ' ', text)


    text = re.sub(r'\s+([,.!?।])', r'\1', text)


    text = re.sub(r'([!?.,।])\1+', r'\1', text)


    text = re.sub(r'([~*_+=-])\1+', r'\1', text)


    text = re.sub(r'\s*([,.!?।])\s*', r'\1 ', text)


    text = text.lower()


    text = text.strip()

    return text
```

```python
normalized_texts = []

for text in noisy_texts:
    normalized_texts.append(normalize_indic_text(text))


normalization_df = pd.DataFrame({
    "Original Text": noisy_texts,
    "Normalized Text": normalized_texts
})

pd.set_option("display.max_colwidth", None)

normalization_df
```

```python
def show_unicode(text):
    print("Text:", text)

    for char in text:
        if not char.isspace():
            print(
                char,
                "->",
                hex(ord(char)),
                unicodedata.name(char, "UNKNOWN")
            )


print("Unicode information for Devanagari:")
show_unicode(devanagari_texts[0])
```

```python
def identify_script(text):
    """
    Rough script identification using Unicode ranges.
    """

    devanagari = 0
    bengali = 0
    latin = 0

    for char in text:
        code = ord(char)

        if 0x0900 <= code <= 0x097F:
            devanagari += 1

        elif 0x0980 <= code <= 0x09FF:
            bengali += 1

        elif ('A' <= char <= 'Z') or ('a' <= char <= 'z'):
            latin += 1

    counts = {
        "Devanagari": devanagari,
        "Bengali": bengali,
        "Latin": latin
    }

    return max(counts, key=counts.get)


print("Script:", identify_script(devanagari_texts[0]))
print("Script:", identify_script(bengali_texts[0]))
```

```python
def script_normalize(text):
    """
    Unicode normalization for Indic scripts.
    NFC combines characters into a standard representation.
    """

    text = unicodedata.normalize("NFC", text)


    text = re.sub(r'\s+', ' ', text)


    text = re.sub(r'\s+([,.!?।])', r'\1', text)

    return text.strip()


normalized_devanagari = [
    script_normalize(text)
    for text in devanagari_texts
]

normalized_bengali = [
    script_normalize(text)
    for text in bengali_texts
]
```

```python
transliterated_devanagari = [
    transliterate(
        text,
        sanscript.DEVANAGARI,
        sanscript.ITRANS
    )
    for text in normalized_devanagari
]


transliterated_bengali = [
    transliterate(
        text,
        sanscript.BENGALI,
        sanscript.ITRANS
    )
    for text in normalized_bengali
]
```

```python
# Unicode-aware tokenizer
tokenizer = nltk.tokenize.RegexpTokenizer(r'\w+|[^\w\s]', flags=re.UNICODE)


def tokenize_text(text):
    return tokenizer.tokenize(text)


parallel_rows = []

for hindi, english in parallel_data:

    hindi_tokens = tokenize_text(hindi)
    english_tokens = tokenize_text(english)

    parallel_rows.append({
        "Hindi Sentence": hindi,
        "English Sentence": english,
        "Hindi Tokens": hindi_tokens,
        "English Tokens": english_tokens,
        "Hindi Length": len(hindi_tokens),
        "English Length": len(english_tokens)
    })


parallel_df = pd.DataFrame(parallel_rows)

parallel_df
```

```python
code_mixed_sentences = [
    "Aaj mera college test hai.",
    "Mujhe Python programming bahut pasand hai.",
    "Kal assignment submit karna hai.",
    "Please mujhe ye file send kar do.",
    "Aaj lecture mein machine learning padha.",
    "Mera laptop bahut slow ho gaya hai.",
    "Tumne project complete kiya kya?",
    "Mujhe ek new notebook chahiye.",
    "Sir ne homework upload karne ko bola.",
    "Main abhi coding practice kar raha hoon.",
    "Aaj presentation ke liye prepare karna hai.",
    "Mujhe ye concept properly samajh nahi aaya."
]

print("Number of code-mixed sentences:", len(code_mixed_sentences))

for i, sentence in enumerate(code_mixed_sentences, 1):
    print(i, sentence)
```

```python
english_words = {
    "college",
    "test",
    "python",
    "programming",
    "assignment",
    "submit",
    "please",
    "file",
    "send",
    "lecture",
    "machine",
    "learning",
    "laptop",
    "slow",
    "project",
    "complete",
    "new",
    "notebook",
    "homework",
    "upload",
    "presentation",
    "prepare",
    "coding",
    "practice",
    "concept",
    "properly"
}


def detect_language(token):
    """
    Simple rule-based language identification.
    This is only for demonstration and is not a full language
    identification model.
    """

    # Devanagari
    if re.search(r'[\u0900-\u097F]', token):
        return "Hindi"

    # English dictionary
    if token.lower() in english_words:
        return "English"

    # Romanized Hindi / unknown Roman word
    if re.fullmatch(r'[A-Za-z]+', token):
        return "Roman-Hindi/Other"

    return "Other"
```

```python
code_mixed_analysis = []

for sentence in code_mixed_sentences:

    tokens = tokenize_text(sentence)

    for token in tokens:

        language = detect_language(token)

        code_mixed_analysis.append({
            "Sentence": sentence,
            "Token": token,
            "Language": language
        })


code_mixed_df = pd.DataFrame(code_mixed_analysis)

code_mixed_df
```

```python
normalized_code_mixed = [
    normalize_indic_text(sentence)
    for sentence in code_mixed_sentences
]


code_mixed_output = pd.DataFrame({
    "Original Code-Mixed Text": code_mixed_sentences,
    "Normalized Text": normalized_code_mixed
})


code_mixed_output
```

```python
def get_token_script(token):

    if re.search(r'[\u0900-\u097F]', token):
        return "Devanagari"

    elif re.fullmatch(r'[A-Za-z]+', token):
        return "Latin"

    else:
        return "Other"


mixed_script_rows = []

for sentence in mixed_devanagari:

    tokens = tokenize_text(sentence)

    for token in tokens:

        mixed_script_rows.append({
            "Sentence": sentence,
            "Token": token,
            "Script": get_token_script(token)
        })


mixed_script_df = pd.DataFrame(mixed_script_rows)

mixed_script_df
```
