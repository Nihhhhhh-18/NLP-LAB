```python
import nltk
import pandas as pd
import numpy as np

from nltk.corpus import brown
from nltk.util import ngrams
from collections import Counter
```

```python

nltk.download('brown')


words = brown.words(categories='news')

print("Total words in news category:", len(words))


words = words[:5000]
```

```python
tokens = []

for word in words:
    word = word.lower()
    if word.isalpha():
        tokens.append(word)

print("Number of tokens after preprocessing:", len(tokens))
print("First 20 tokens:")
print(tokens[:20])
```

```python

total_tokens = len(tokens)


vocabulary = set(tokens)
vocabulary_size = len(vocabulary)

print("Total Tokens:", total_tokens)
print("Vocabulary Size:", vocabulary_size)
```

```python

unigrams = list(ngrams(tokens, 1))
bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))

print("Number of Unigrams:", len(unigrams))
print("Number of Bigrams:", len(bigrams))
print("Number of Trigrams:", len(trigrams))

print("\nSample Unigrams:")
print(unigrams[:10])

print("\nSample Bigrams:")
print(bigrams[:10])

print("\nSample Trigrams:")
print(trigrams[:10])
```

```python

unigram_freq = Counter(unigrams)
bigram_freq = Counter(bigrams)
trigram_freq = Counter(trigrams)

print("\nTop 10 Most Frequent Unigrams:")
for item, count in unigram_freq.most_common(10):
    print(item, ":", count)

print("\nTop 10 Most Frequent Bigrams:")
for item, count in bigram_freq.most_common(10):
    print(item, ":", count)

print("\nTop 10 Most Frequent Trigrams:")
for item, count in trigram_freq.most_common(10):
    print(item, ":", count)
```

```python
import nltk
import math

from nltk.corpus import gutenberg
from collections import Counter


nltk.download('gutenberg')


print("Available Gutenberg texts:")
print(gutenberg.fileids())
```

```python

words = gutenberg.words('austen-emma.txt')


tokens = []

for word in words:
    word = word.lower()
    if word.isalpha():
        tokens.append(word)

print("Total tokens:", len(tokens))
print("First 20 tokens:")
print(tokens[:20])
```

```python
unigram_list = list(ngrams(tokens, 1))
unigram_counts = Counter(unigram_list)

bigram_list = list(ngrams(tokens, 2))
bigram_counts = Counter(bigram_list)

vocab_size = len(unigram_counts)

print("Vocabulary Size:", vocab_size)
print("Number of unique unigrams:", len(unigram_counts))
print("Number of unique bigrams:", len(bigram_counts))

print("\nTop 10 Unigrams:")
for word, count in unigram_counts.most_common(10):
    print(word, ":", count)

print("\nTop 10 Bigrams:")
for pair, count in bigram_counts.most_common(10):
    print(pair, ":", count)
```

```python

def bigram_probability(previous_word, current_word):

    bigram_count = bigram_counts[(previous_word, current_word)]
    previous_count = unigram_counts[previous_word]

    probability = (bigram_count + 1) / (previous_count + vocab_size)

    return probability
```

```python

def sentence_probability(sentence):

    words = sentence.lower().split()


    words = [word for word in words if word.isalpha()]

    probability = 1.0

    for i in range(1, len(words)):

        prob = bigram_probability(words[i-1], words[i])
        probability = probability * prob

    return probability



test_sentences = [
    "she was very happy",
    "the man was walking",
    "he went to the house"
]

print("Sentence Scores:\n")

for sentence in test_sentences:
    score = sentence_probability(sentence)

    print("Sentence:", sentence)
    print("Probability:", score)
    print()
```

```python
print("""
Observation:

Without smoothing, a bigram that was not present in the training corpus
would receive probability zero.

If even one bigram in a sentence has probability zero, the probability
of the complete sentence would also become zero.

Add-one smoothing solves this problem by adding 1 to every bigram count.
Therefore, even unseen bigrams receive a small non-zero probability.

Smoothing helps language models handle new or unseen word combinations.
""")
```

```python
import nltk

from nltk.corpus import brown


nltk.download('brown')
nltk.download('universal_tagset')


tagged_sentences = brown.tagged_sents(
    categories='news',
    tagset='universal'
)

print("Total tagged sentences:", len(tagged_sentences))

print("\nFirst tagged sentence:")
print(tagged_sentences[0])
```

```python

split = int(0.8 * len(tagged_sentences))

train_data = tagged_sentences[:split]
test_data = tagged_sentences[split:]

print("Training sentences:", len(train_data))
print("Testing sentences:", len(test_data))
```

```python
from nltk.tag import HiddenMarkovModelTrainer


trainer = HiddenMarkovModelTrainer()


hmm_tagger = trainer.train_supervised(train_data)

print("HMM tagger trained successfully.")
```

```python

test_sentences = [
    "The dog runs quickly",
    "The student reads a book",
    "She likes the beautiful garden",
    "The company announced new plans",
    "John bought a new car"
]

print("Predicted POS Tags:\n")

for sentence in test_sentences:

    words = sentence.lower().split()

    predicted_tags = hmm_tagger.tag(words)

    print("Sentence:", sentence)
    print("Tags:", predicted_tags)
    print()
```

```python

accuracy = hmm_tagger.accuracy(test_data)

print("HMM POS Tagging Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")
```

```python
print("""
HMM Components in POS Tagging:

1. Hidden States:
   The POS tags are the hidden states.
   Examples: NOUN, VERB, ADJ, ADV.

2. Observations:
   The actual words in the sentence are observations.
   For example:
   "The dog runs"
   The observations are The, dog and runs.

3. Transition Probability:
   It represents the probability of moving from one POS tag
   to another POS tag.

   Example:
   P(VERB | NOUN)

4. Emission Probability:
   It represents the probability of a particular word being
   generated by a particular POS tag.

   Example:
   P("runs" | VERB)

5. Initial Probability:
   It represents the probability of a POS tag occurring
   at the beginning of a sentence.

The HMM uses these probabilities to find the most likely sequence
of POS tags for a given sequence of words.
""")
```

```python
from sklearn.datasets import fetch_20newsgroups


categories = ['sci.space', 'comp.graphics']

newsgroups = fetch_20newsgroups(
    subset='train',
    categories=categories,
    remove=('headers', 'footers', 'quotes')
)

print("Categories:", newsgroups.target_names)
print("Number of documents:", len(newsgroups.data))

print("\nFirst document:")
print(newsgroups.data[0][:1000])
```

```python
import re
from collections import Counter


all_text = " ".join(newsgroups.data)


words = re.findall(r'\b[a-zA-Z]{3,}\b', all_text.lower())


stop_words = {
    'the', 'and', 'for', 'that', 'this', 'with',
    'from', 'have', 'are', 'was', 'were', 'you',
    'not', 'but', 'they', 'can', 'will', 'would',
    'about', 'there', 'their', 'what', 'which',
    'when', 'where', 'who', 'how', 'has', 'had',
    'been', 'also', 'one', 'all', 'out', 'more',
    'some', 'than', 'into', 'its', 'our', 'your',
    'just', 'like', 'use', 'using', 'does', 'did',
    'get', 'may', 'such', 'any', 'only', 'other'
}


domain_words = []

for word in words:
    if word not in stop_words:
        domain_words.append(word)


domain_frequency = Counter(domain_words)

print("Top 20 Frequent Domain-Specific Terms:\n")

for word, count in domain_frequency.most_common(20):
    print(word, ":", count)
```

```python
for category in categories:

    print("\n" + "=" * 50)
    print("DOMAIN:", category)
    print("=" * 50)


    category_data = fetch_20newsgroups(
        subset='train',
        categories=[category],
        remove=('headers', 'footers', 'quotes')
    )

    text = " ".join(category_data.data)

    words = re.findall(r'\b[a-zA-Z]{3,}\b', text.lower())

    filtered_words = [
        word for word in words
        if word not in stop_words
    ]

    counts = Counter(filtered_words)

    print("Top 10 terms:")

    for word, count in counts.most_common(10):
        print(word, ":", count)
```
